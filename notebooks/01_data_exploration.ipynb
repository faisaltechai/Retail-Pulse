```python
cancelled = df['InvoiceNo'].astype(str).str.startswith("C")
print(f'The total number of cancelled orders is: {cancelled.sum()}')
```

```python
df[cancelled]
```

```python
df[cancelled].head()
```

```python
cancelled_percentage = cancelled.sum() /  len(cancelled)* 100
```

```python
print(f"The percentage of cancelled order is : {cancelled_percentage:.2f}%")
```

```python
cancelled_df = df[cancelled].copy()

cancelled_revenue = cancelled_df['Quantity'] * cancelled_df['UnitPrice']

print(f"Cancelled revenue: £{cancelled_revenue.sum()}")
```

```python
print(f"Cancelled quantity: {cancelled_df['Quantity'].sum().__abs__()}")
```
